# Logistic Regression
Last week, we introduced the framework of Generalised Linear Models (GLMs). This week, we will see our first example of a specific GLM in the form of *logistic regression*. This is one of the most commonly applied GLMs and can be used under two different circumstances. The first is where the outcome variable is *binary* and the second is where the outcome is the *count of successes* in a sequence of binary trials. For instance, in a memory experiment we could have a pass/fail threshold dictated by the number of items remembered. Each subject would therefore either pass or fail, making the outcome variable binary. Alternatively, we could count the number of correctly remembered items for each subject. In this case, the outcome variables would be a *count of successes* across the memory trials. Logistic regression can be used in both cases. We will refer to these as *binary logistic regression* and *binomial logistic regression*. This lesson will focus primarily on binary logistic regression, with binominal logistic regression demonstrated during the associated workshop.
