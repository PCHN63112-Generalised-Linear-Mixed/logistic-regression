# Logistic Regression
...

Logistic regression can be used under two different circumstances. One is where the outcome variable is *binary* and the assumed marginal distribution is a Bernoulli distribution. The second is where the outcome is the count of "successes" in $n$ Bernoulli trials and the assumed marginal distribution is a binomial distribution. This is a direct consequence of the fact that the Bernoulli is simply a special case of the binomial and thus logistic regression can be used under either circumstance. In this lesson, we refer to these as *binary logistic regression* and *binomial logistic regression*, and will provides examples of both.
