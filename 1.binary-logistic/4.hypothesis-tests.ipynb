## Analysis of Deviance

```r
mod.1 <- glm(disease ~ age, family=binomial(link=logit), data=cleveland)
mod.2 <- glm(disease ~ poly(age,2), family=binomial(link=logit), data=cleveland)
mod.3 <- glm(disease ~ poly(age,3), family=binomial(link=logit), data=cleveland)

anova(mod.1,mod.2,mod.3)
```
