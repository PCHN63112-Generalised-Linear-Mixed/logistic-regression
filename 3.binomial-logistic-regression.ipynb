# Binomial Logistic Regression
... One example would be aggregating a number of trials of a memory task for each subject. In this case, each individual trial would be binary (0 = forgotten, 1 = remembered), but we could sum all the trials together to give a total for each subject 

$$
y_{i} = \sum_{j=1}^{n_{i}} s_{ij},
$$

where $s_{ij}$ is the binary score for subject $i$ on trial $j$. Our outcome would then be a *count* of the number of "successes" across $n_{i}$ trials. 

## Distribution and Link Function

### The Binomial Distribution

$$
y_{i} \sim \text{Binom}\left(\pi_{i},n_{i}\right)
$$

Notice here that *both* the probability of success and the number of trials is allowed to vary across observations. We assume that $n_{i}$ is known so that only $\pi_{i}$ is estimated. However, $n_{i}$ does not have to be the same for each observation. This flexibility is particularly useful when aggregating trials, become some subjects may not provide a response on every trial and thus the number of *complete* trials will differ across subjects. Conceptually, we assume that two subjects with the same predictor variables values have the same probability of success, irrespective of the number of trials.
